```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import filters, segmentation, color
from skimage.segmentation import active_contour
from skimage.filters import sobel, prewitt
from google.colab import files
from io import BytesIO
from PIL import Image

uploaded = files.upload()

image = Image.open(BytesIO(uploaded[list(uploaded.keys())[0]]))
image = np.array(image)

plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.show()

gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.title("Thresholding")
plt.axis("off")
plt.show()

gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.title("Thresholding")
plt.axis("off")
plt.show()

hsv = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)

lower = np.array([0, 50, 50])
upper = np.array([10, 255, 255])

mask = cv2.inRange(hsv, lower, upper)
result = cv2.bitwise_and(image, image, mask=mask)

plt.imshow(result)
plt.title("Color Segmentation")
plt.axis("off")
plt.show()

edges_canny = cv2.Canny(gray, 100, 200)

plt.imshow(edges_canny, cmap='gray')
plt.title("Canny Edge")
plt.axis("off")
plt.show()

edges_sobel = sobel(gray)

plt.imshow(edges_sobel, cmap='gray')
plt.title("Sobel Edge")
plt.axis("off")
plt.show()

edges_prewitt = prewitt(gray)

plt.imshow(edges_prewitt, cmap='gray')
plt.title("Prewitt Edge")
plt.axis("off")
plt.show()

gray_float = color.rgb2gray(image)

s = np.linspace(0, 2*np.pi, 400)
r = 100 + 100*np.sin(s)
c = 100 + 100*np.cos(s)
init = np.array([r, c]).T

snake = active_contour(
    filters.gaussian(gray_float, 3),
    init,
    alpha=0.015,
    beta=10,
    gamma=0.001
)

fig, ax = plt.subplots()
ax.imshow(gray_float, cmap='gray')
ax.plot(init[:, 1], init[:, 0], '--r')
ax.plot(snake[:, 1], snake[:, 0], '-b')
ax.set_title("Active Contour")
ax.axis('off')
plt.show()
```
